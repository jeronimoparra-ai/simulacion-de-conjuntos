```python
#  Sum all the numbers in the list xs, 

def sum_all(xs):
    total = 0
    for x in xs:
        total = total + x
    return total 


```

```python
# lista de numeros = [1, 2, 3, 4, 5]
lista_de_numeros = [1, 2, 3, 4, 5]
```

```python
sum_all(lista_de_numeros)  # sum_all es una función fructifera porque devuelve un valor
```

```python
suma = sum_all(lista_de_numeros)
```

```python
def sum_to(n):
    """ Return the sum of 1+2+3 ... n """
    ss  = 0 # Initialize sum to zero
    v = 1 # The current value to add
    while v <= n:
        ss = ss + v
        v = v + 1
    return ss
```

```python
sum_to(4) 
```

```python
def seq3np1(n):
    """ Print the 3n+1 sequence from n,
        terminating when it reaches 1.
    """
    while n != 1:
        print(n, end=", ")
        if n % 2 == 0:        # n is even
            n = n // 2       # Use integer division
        else:                 # n is odd
            n = n * 3 + 1  # Compute 3n+1
    print(n, end=".\n")  # Print the final 1
```

```python
seq3np1(6)
```
